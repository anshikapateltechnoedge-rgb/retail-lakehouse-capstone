```python
# Job task 1: load new files from the landing volume into bronze
src = "/Volumes/workspace/capstone/landing/"
spark.readStream.format("cloudFiles").option("cloudFiles.format", "csv").option("header", True).option("cloudFiles.inferColumnTypes", True).option("cloudFiles.schemaLocation", src + "_schemas/orders").load(src + "orders/").selectExpr("*", "_metadata.file_name AS source_file", "current_timestamp() AS ingested_at").writeStream.option("checkpointLocation", src + "_checkpoints/bronze_orders").trigger(availableNow=True).toTable("workspace.capstone.bronze_orders").awaitTermination()
print("bronze rows:", spark.table("workspace.capstone.bronze_orders").count())
```
